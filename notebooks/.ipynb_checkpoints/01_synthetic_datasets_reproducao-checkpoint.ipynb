## Desafios e correções durante a reprodução

Durante a execução, alguns ajustes foram necessários em relação à lógica 
inicialmente descrita:
- O parâmetro `size` não existe em `empty_add()`; o correto é `radius`
- `constraint_move_up()` só funciona no contexto da interface gráfica; 
  a solução foi criar a constraint "Child Of" antes da "Track To", 
  garantindo a ordem correta na pilha sem precisar reordenar
- Cena renderizava em preto: era necessário adicionar objeto visível, 
  luz (Sun), definir a câmera como ativa (`scene.camera`), e desativar 
  transparência de fundo (`film_transparent = False`)
